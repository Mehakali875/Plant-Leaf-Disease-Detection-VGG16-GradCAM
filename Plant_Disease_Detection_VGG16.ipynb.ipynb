**Libraries**

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import VGG16
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping

import matplotlib.pyplot as plt
import numpy as np
import cv2
import seaborn as sns

from sklearn.metrics import confusion_matrix, classification_report
import os
```

TensorFlow is used for Deep Learning, NumPy for arrays, Matplotlib for graphs, OpenCV for image processing, and Scikit-learn for evaluation metrics.

**Dataset Paths**


```python
train_path = "/content/dataset/Plant_Disease_Project/train"
val_path   = "/content/dataset/Plant_Disease_Project/val"
test_path  = "/content/dataset/Plant_Disease_Project/test"
```

The code stores the locations of train, validation, and test folders in variables so they can be easily used later in the project.

**Image Preprocessing and Augmentation**

```python
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True
)

val_datagen = ImageDataGenerator(
    rescale=1./255
)

test_datagen = ImageDataGenerator(
    rescale=1./255
)
```

Images are resized and normalized. Data augmentation randomly rotates, zooms, and flips images to improve model performance and reduce overfitting.

**Load Training, Validation, and Test Data**

```python
datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

# Training Data
train_data = datagen.flow_from_directory(
    train_path,
    target_size=(224,224),
    batch_size=32,
    class_mode='categorical',
    subset='training'
)

# Validation Data
val_data = datagen.flow_from_directory(
    train_path,
    target_size=(224,224),
    batch_size=32,
    class_mode='categorical',
    subset='validation'
)

# Test Data
test_data = datagen.flow_from_directory(
    test_path,
    target_size=(224,224),
    batch_size=32,
    class_mode='categorical',
    shuffle=False
)
```

The images are loaded from folders and converted into batches so the Deep Learning model can process them efficiently during training.

**Dataset Class Names**

```python
print("Class Names:")
print(train_data.class_indices)
```

The code shows all categories available in the dataset such as healthy or diseased plant leaves.

**VGG16 Transfer Learning Model**

```python
base_model = VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(224,224,3)
)

# Freeze pretrained layers
for layer in base_model.layers:
    layer.trainable = False
```

**Building Deep Learning Model**

```python
model = models.Sequential([

    base_model,

    layers.Flatten(),

    layers.Dense(256, activation='relu'),

    layers.Dropout(0.5),

    layers.Dense(4, activation='softmax')

])
```

**Compile Model**

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

The Adam optimizer is used for learning, categorical crossentropy is used as the loss function, and accuracy is used as the performance metric.

**Model**

```python
model.summary()
```

The output shows all layers, parameters, and shapes used in the Deep Learning model.

**Early Stopping**

```python
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)
```

Prevents overfitting – if the model stops improving on unseen data, we stop early to save time and keep the best model.

**Model Training**

```python
model = models.Sequential([
    base_model,
    layers.Flatten(),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(4, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Explicitly build the model after re-definition
# (None for batch size, 224, 224 for image size, 3 for channels)
model.build(input_shape=(None, 224, 224, 3))
```

```python
# Training start
history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=10,
    callbacks=[early_stop]
)
```

 Trains the model using training data and validates on validation data.


*   train_data – augmented training images.

*   validation_data – validation images.


*   epochs=10 – maximum 10 passes through the data (early stopping may end sooner).
*   callbacks=[early_stop] – stops early if needed.


**Save Trained Model**

```python
model.save("plant_disease_model.h5")

print("Model Saved Successfully!")
```

Saves the model weights and architecture to an HDF5 file for later use (deployment or further evaluation).

Similar to accuracy plot, but for loss values. Loss should decrease over time for both sets.

**Evaluate on Test Data**

```python
loss, accuracy = model.evaluate(test_data)
print(f"\nTest Accuracy: {accuracy*100:.2f}%")
```

his tells us how well our model generalizes to new leaf images.

**Predictions**

```python
predictions = model.predict(test_data)
predicted_classes = np.argmax(predictions, axis=1)
true_classes = test_data.classes
class_labels = list(test_data.class_indices.keys())
```

Computes the model's predicted classes for all test images and stores the true labels for comparison.

**Confusion Matrix**

```python
cm = confusion_matrix(true_classes, predicted_classes)

plt.figure(figsize=(8,6))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_labels,
    yticklabels=class_labels
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

Visualizes correct vs incorrect predictions per class. Diagonal cells show correct predictions, off‑diagonals show where the model confused classes.

**Classification Report**


```python
print("\nClassification Report:\n")

print(classification_report(
    true_classes,
    predicted_classes,
    target_names=class_labels
))
```

Prints precision, recall, F1‑score, and support for each class. These metrics give a detailed view of model performance per disease category.



**Predict_image Function**

```python
def predict_image(img_path):

    img = tf.keras.utils.load_img(
        img_path,
        target_size=(224,224)
    )

    img_array = tf.keras.utils.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    img_array = img_array / 255.0
    prediction = model.predict(img_array)
    predicted_class = class_labels[np.argmax(prediction)]
    confidence = np.max(prediction) * 100

    plt.imshow(img)
    plt.title(f"Prediction: {predicted_class}\nConfidence: {confidence:.2f}%")
    plt.axis('off')
    plt.show()

    return predicted_class
```

Allows testing the model on a single image file. It loads, resizes, normalizes, predicts, and displays the result with confidence.

**Grad-CAM Inspired Disease Region Visualization**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def simple_heatmap(img_path):

    # Read image
    img = cv2.imread(img_path)
    img = cv2.resize(img, (224,224))

    # Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Apply heatmap
    heatmap = cv2.applyColorMap(gray, cv2.COLORMAP_JET)

    # Overlay heatmap
    output = cv2.addWeighted(img, 0.6, heatmap, 0.4, 0)

    # Show original image
    plt.figure(figsize=(6,6))
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")
    plt.show()

    # Show heatmap image
    plt.figure(figsize=(6,6))
    plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
    plt.title("Disease Region Visualization")
    plt.axis("off")
    plt.show()
```

Creates a basic heatmap overlay to highlight areas of the image that might indicate disease. Although not a true Grad‑CAM, it gives a visual clue about which regions are “hot” (possibly diseased) based on grayscale intensity.
